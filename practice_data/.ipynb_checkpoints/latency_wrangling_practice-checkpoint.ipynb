# Latency data-wrangling practice

Every exercise here comes from a question you've actually asked about TT latency data: stitching exports, pairing order events, bucketing by session, matching rejects, joining market data, and getting the plot to say what you want. The data is fake but shaped like the real exports, quirks included. It also has a few **planted effects** for you to find.

**How to use it**
1. Run the setup cell. If `practice_data/` is missing, it runs `make_practice_data.py` to create it (about 10 s).
2. For each exercise, write your code in the empty cell, then run the `# check` cell under it. Checks compare your numbers with `practice_data/answers.json`, which the generator wrote from the ground truth.
3. Open the **Hints** only after a real attempt. They name tools to look up, not answers.
4. Plot exercises have no check. Compare yours with the target and make it say the same thing.
5. To start again with different numbers, run `python make_practice_data.py --seed 11`. The checks follow the new answer key.

The exercises name the variables they expect (`submit`, `orders`, `life` …) because later checks use them.

---

## The data

| path | what it is |
|---|---|
| `submit_data/DD_MM_<sess>[_patch].csv` | Order submissions: `NewOrderSingle` (our send, no latency), `ExecutionReport/New` (exchange ack, with `Exch Latency (μs)`), `ExecutionReport/Trade` (fills). Split into morn/arvo files, plus some patch files. |
| `cancel_data/DD_MM_cans_<sess>.csv` | `CancelRequest` (our cancel, no latency) and `ExecutionReport/Canceled` (exchange ack, with latency). |
| `rejects_*.csv` | Cancel rejects: `Message Type = Reject`, reason text in `Message`. |
| `mki/mki_21_10.csv` | Another instrument (MKI): ExecutionReports only, from an export done with TT's **UTC** display option. |
| `market_data/krx_activity.csv` | Exchange-wide messages per second (`utc_time`, `msgs`). |
| `market_data/s50_tob_DD_MM.csv` | S50 top of book for one day: `recvTime` / `exchTime` in **ns since 1970, UTC**, `bidPx`, `askPx`, quantities. |

**TT facts you'll need**
- `Time` is when TT sent or received the message, in Sydney time unless stated otherwise. It's logged to the millisecond.
- `Exch Latency (μs)` = TT sends the order/cancel → TT receives the exchange's ack. It only appears on the ack row.
- Every message for one order shares a `TTOrderID`.

Treat the exports as you'd treat real ones: look before you trust.

### D1  "Why do I only get one row?"

This builds one row per New → Canceled order on 12 Oct (morning). Expect a couple of thousand rows.

**Think:** Once it works, count how many orders hit the `continue`, and why each one did.

```python
sub = pd.read_csv(DATA / 'submit_data' / '12_10_morn.csv')
can = pd.read_csv(DATA / 'cancel_data' / '12_10_cans_morn.csv')
both = pd.concat([sub, can])
both = both[both['Contract'] == 'S50 Nov26']

results = []
for order_id, g in both.groupby('TTOrderID'):
    new = g[g['ExecType'] == 'New']
    canc = g[g['ExecType'] == 'Canceled']
    if len(new) != 1 or len(canc) != 1:
        continue
    new_ms = new[LAT].iloc[0] / 1000
results.append({'TTOrderID': order_id, 'Time': canc['Time'].iloc[0], 'New': new_ms})

results = pd.DataFrame(results)
print(len(results), 'orders')
```

```python
# check
assert len(results) > 1000, f'only {len(results)} rows'; print('✓ D1')
```
