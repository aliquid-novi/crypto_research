# Part 1 — Abstraction and polymorphism in general

## 1.1 What problem is abstraction solving?

Abstraction separates **what a thing promises to do** from **how it does it**. Callers depend
on the promise, implementations vary underneath.

The payoff is not code reuse. It is that you can add a fifth fee model, a third venue or a
second PnL convention without editing the code that consumes them. If you are never going to
have a second implementation, an abstraction costs you indirection and buys nothing.

## 1.2 Python already has polymorphism without any of this

Start here, because it is the honest baseline. Python calls methods by name at runtime. Two
unrelated classes with the same method are already interchangeable.

````python
class SpotFX:
    def notional(self, qty, price):
        return abs(qty) * price


class LinearPerp:
    def __init__(self, multiplier):
        self.multiplier = multiplier

    def notional(self, qty, price):
        return abs(qty) * price * self.multiplier


# Same call site, different behaviour. No base class, no interface, no imports.
for inst in (SpotFX(), LinearPerp(0.01)):
    print(f"{type(inst).__name__:<12} {inst.notional(10, 60_000.0):>12,.2f}")
````

That is **duck typing**: the object is acceptable because it has the method, not because of
what it inherits from.

So what does `abc` add? Three things:

1. **Enforcement** — a subclass that forgets a method fails at construction, not at 3am when
   that branch is first hit.
2. **Documentation** — the base class is a written list of what an implementation owes you.
3. **A place to put shared logic** — the invariant parts of a workflow live in the base, the
   varying parts are left abstract.

Point 3 is usually the one that earns its keep.

## 1.3 `ABC` and `@abstractmethod`

````python
from abc import ABC, abstractmethod


class Instrument(ABC):
    @abstractmethod
    def notional(self, qty, price):
        # No body needed. The contract is the signature.
        ...

    @abstractmethod
    def pnl(self, entry, exit_price, qty):
        ...


class Incomplete(Instrument):
    def notional(self, qty, price):
        return abs(qty) * price
    # pnl is missing


try:
    Incomplete()
except TypeError as e:
    print("TypeError:", e)
````

The check happens at **instantiation**, not at class definition and not at call time. That is
the whole point: you find out when the object is built, which in a trading system is at
startup rather than mid-session.

Note also that `Instrument()` itself cannot be constructed.

````python
try:
    Instrument()
except TypeError as e:
    print("TypeError:", e)
````

## 1.4 The trap: `@abstractmethod` only works with `ABCMeta`

`@abstractmethod` on its own is just a marker attribute. Without `ABC` (or
`metaclass=ABCMeta`) in the bases, **nothing enforces it**.

````python
class LooksAbstract:  # plain object — no ABC in the bases
    @abstractmethod
    def must_implement(self):
        ...


obj = LooksAbstract()   # constructs happily
print("constructed:", obj)
print("marker present:", LooksAbstract.must_implement.__isabstractmethod__)
print("returns:", obj.must_implement())   # returns None, silently
````

This one bites people who copy a decorator without the base class. If a subclass is silently
instantiating when you expected a `TypeError`, check the bases first.

## 1.5 Abstract properties, classmethods, staticmethods

Decorator order matters: `@abstractmethod` goes **innermost** (applied first).

````python
class Venue(ABC):
    @property
    @abstractmethod
    def name(self):
        ...

    @property
    @abstractmethod
    def taker_bps(self):
        ...

    @classmethod
    @abstractmethod
    def from_config(cls, cfg):
        ...

    @staticmethod
    @abstractmethod
    def normalise_symbol(raw):
        ...


class VenueA(Venue):
    name = "venue-a"            # a plain class attribute satisfies an abstract property
    taker_bps = 4.5

    @classmethod
    def from_config(cls, cfg):
        return cls()

    @staticmethod
    def normalise_symbol(raw):
        return raw.upper().replace("-", "")


v = VenueA.from_config({})
print(v.name, v.taker_bps, VenueA.normalise_symbol("btc-perp"))
````

Notice `name = "venue-a"` satisfies `@property @abstractmethod name`. The ABC machinery checks
that *the attribute name exists* on the class, not that it is a property. That is a feature
when you want constants, and a hazard if you assumed the base could force a computed value.

## 1.6 The template method: shared skeleton, abstract holes

This is the pattern that justifies most abstract base classes in trading code. The base owns
the **sequence and the invariants**; subclasses fill in only what varies.

````python
class RiskCheckedOrderFlow(ABC):
    def __init__(self, max_position):
        self.max_position = max_position
        self.position = 0.0
        self.rejected = 0

    # --- concrete: this sequence is identical for every strategy, so nobody can skip a step
    def submit(self, signal, book):
        order = self.build_order(signal, book)
        if order is None:
            return None
        order = self.apply_limits(order)
        if not self.within_risk(order):
            self.rejected += 1
            return None
        self.position += order["qty"]
        return order

    # --- abstract: the part that actually differs between strategies
    @abstractmethod
    def build_order(self, signal, book):
        ...

    # --- hook: a default subclasses may override, but do not have to
    def apply_limits(self, order):
        order["qty"] = max(-5.0, min(5.0, order["qty"]))
        return order

    def within_risk(self, order):
        return abs(self.position + order["qty"]) <= self.max_position


class MomentumFlow(RiskCheckedOrderFlow):
    def build_order(self, signal, book):
        if abs(signal) < 0.5:
            return None
        return {"qty": 10.0 if signal > 0 else -10.0, "px": book["mid"]}


f = MomentumFlow(max_position=8.0)
for s in (0.9, 0.1, 0.8, 0.8):
    print(s, "->", f.submit(s, {"mid": 100.0}), "pos:", f.position, "rejected:", f.rejected)
````

The clamp to ±5 and the position check cannot be forgotten by a new strategy, because the new
strategy never writes the sequence. That is the difference between an interface and a
framework: an interface says what you may call, a template method says what will happen.

**Where it goes wrong:** if the fixed sequence in `submit` is not actually universal, every
subclass starts overriding `submit` to work around it, and you have inheritance with none of
the guarantees. When you see that happening, the sequence was a guess, not an invariant.

## 1.7 `register()` — virtual subclasses

`ABC.register` makes `isinstance` return `True` for a class you cannot edit (third-party, or
generated). It is a bare assertion: **nothing is checked**.

````python
class Quotable(ABC):
    @abstractmethod
    def best_bid(self):
        ...


class ThirdPartyBook:          # imagine this comes from a vendor package
    def best_bid(self):
        return 99.5


class LyingBook:               # does not implement best_bid at all
    pass


Quotable.register(ThirdPartyBook)
Quotable.register(LyingBook)

print(isinstance(ThirdPartyBook(), Quotable))   # True
print(isinstance(LyingBook(), Quotable))        # also True — register checks nothing
try:
    LyingBook().best_bid()
except AttributeError as e:
    print("AttributeError:", e)
````

So `register` buys you `isinstance` compatibility and nothing else. Use it to adapt foreign
types at a boundary, never as a substitute for implementing the contract.

## 1.8 `__subclasshook__` — structural `isinstance`

If you want `isinstance` to succeed for *anything shaped right*, implement the hook. This is
how `collections.abc.Iterable` works.

````python
class Markable(ABC):
    @abstractmethod
    def mark(self, symbol):
        ...

    @classmethod
    def __subclasshook__(cls, C):
        if cls is Markable:
            return any("mark" in B.__dict__ for B in C.__mro__)
        return NotImplemented


class NeverHeardOfMarkable:
    def mark(self, symbol):
        return 100.0


print(issubclass(NeverHeardOfMarkable, Markable))
print(isinstance(NeverHeardOfMarkable(), Markable))
````

## 1.9 `Protocol` — structural typing without inheritance

`typing.Protocol` gives you the interface as a *static* contract. Implementations do not
import or inherit from it, which matters when the thing you are describing lives in someone
else's package.

````python
from typing import Protocol, runtime_checkable


@runtime_checkable
class PriceSource(Protocol):
    def mark(self, symbol: str) -> float: ...


class LastTradeMark:
    def __init__(self, trades):
        self.trades = trades

    def mark(self, symbol):
        return self.trades[symbol][-1]


class MidMark:
    def __init__(self, book):
        self.book = book

    def mark(self, symbol):
        bid, ask = self.book[symbol]
        return (bid + ask) / 2.0


def pv(position: float, symbol: str, src: PriceSource) -> float:
    return position * src.mark(symbol)


print(pv(3.0, "BTC", LastTradeMark({"BTC": [59_900.0, 60_100.0]})))
print(pv(3.0, "BTC", MidMark({"BTC": (59_990.0, 60_010.0)})))
print(isinstance(MidMark({}), PriceSource))
````

````python
# Caveat: runtime_checkable only checks that the *name* exists. Signatures are ignored.
class WrongSignature:
    def mark(self):     # takes no symbol
        return 1.0


print(isinstance(WrongSignature(), PriceSource))   # True — and it will blow up when called
try:
    pv(1.0, "BTC", WrongSignature())
except TypeError as e:
    print("TypeError:", e)
````

**ABC or Protocol?**

| | ABC | Protocol |
|---|---|---|
| Implementations must inherit | yes | no |
| Can carry shared concrete logic | yes | not really |
| Catches a missing method | at construction | at type-check time (mypy/pyright) |
| Good for | your own hierarchy, template methods | describing what you accept, adapting foreign types |

Rule of thumb: **Protocol for what a function accepts, ABC for a family you own and want to
share behaviour across.**

## 1.10 Four kinds of polymorphism worth naming

1. **Subtype** — `Instrument` subclasses, dispatch on the receiver.
2. **Duck typing** — anything with the method, no declared relationship.
3. **Ad hoc / overloading** — dispatch on the *argument* type (`functools.singledispatch`).
4. **Parametric** — one implementation over many types (generics in Python, templates in C++).

Subtype dispatch is on the receiver. When the thing that varies is the *message* rather than
the handler, single dispatch on argument type reads better than a chain of `isinstance`.

````python
from dataclasses import dataclass
from functools import singledispatch


@dataclass
class Fill:
    symbol: str
    qty: float
    price: float


@dataclass
class Funding:
    symbol: str
    rate: float


@dataclass
class Heartbeat:
    ts: int


@singledispatch
def handle(event):
    raise NotImplementedError(f"no handler for {type(event).__name__}")


@handle.register
def _(event: Fill):
    return f"fill  {event.qty:+.2f} {event.symbol} @ {event.price}"


@handle.register
def _(event: Funding):
    return f"fund  {event.symbol} {event.rate:+.4%}"


@handle.register
def _(event: Heartbeat):
    return "hb"


for e in (Fill("BTC", 1.0, 60_000.0), Funding("BTC", 0.0001), Heartbeat(1)):
    print(handle(e))

try:
    handle("not an event")
except NotImplementedError as e:
    print("NotImplementedError:", e)
````

## 1.11 Operator polymorphism (dunder methods)

Same idea, applied to syntax. Useful for value types you add up a lot.

````python
@dataclass(frozen=True)
class PnL:
    realized: float = 0.0
    unrealized: float = 0.0
    fees: float = 0.0
    funding: float = 0.0

    def __add__(self, other):
        if not isinstance(other, PnL):
            return NotImplemented          # let Python try other.__radd__
        return PnL(self.realized + other.realized,
                   self.unrealized + other.unrealized,
                   self.fees + other.fees,
                   self.funding + other.funding)

    # No __radd__ here. It would not rescue sum() anyway: sum() starts from int 0, so
    # 0 + PnL(...) calls PnL.__radd__(0), which would hit the isinstance check and fail.

    @property
    def total(self):
        return self.realized + self.unrealized - self.fees + self.funding

    def __str__(self):
        return (f"real {self.realized:>10,.2f} | unreal {self.unrealized:>10,.2f} | "
                f"fees {self.fees:>8,.2f} | fund {self.funding:>8,.2f} "
                f"|| total {self.total:>10,.2f}")


book = [PnL(realized=120.0, fees=3.1), PnL(unrealized=-40.0, funding=1.2), PnL(realized=15.0)]
print(sum(book, PnL()))
````

Returning `NotImplemented` (not raising) is the correct signal for "I do not know how to add
that" — Python then tries the reflected operation before raising `TypeError`. Because
`sum()` starts from the integer `0`, the start value `PnL()` is passed explicitly.

## 1.12 Liskov: the thing ABCs do **not** give you

An ABC enforces that a method *exists*. It cannot enforce that the method *behaves*. The last
two classes below satisfy the ABC, and both break callers.

````python
class Order(ABC):
    @abstractmethod
    def cancel(self):
        ...


class Resting(Order):
    def cancel(self):
        return True


class ImmediateOrCancel(Order):
    def cancel(self):
        # Satisfies the ABC. Destroys any caller that loops over orders and cancels them.
        raise NotImplementedError("IOC orders cannot be cancelled")


class QuietlyBroken(Order):
    def cancel(self):
        return None      # caller checks `if order.cancel():` and silently treats it as failed


for o in (Resting(), ImmediateOrCancel(), QuietlyBroken()):
    try:
        print(f"{type(o).__name__:<18} -> {o.cancel()}")
    except NotImplementedError as e:
        print(f"{type(o).__name__:<18} -> raised: {e}")
````

Three failure modes that pass every mechanical check:

- **Strengthened precondition** — subclass rejects inputs the base accepted.
- **Weakened postcondition** — subclass returns less, or a different type, or `None`.
- **New exceptions** — subclass raises something the base never documented.

The defence is a **contract test**: one test function written against the base, run against
every implementation. Part 2 has a worked example.

> ### C++ mapping — Part 1
>
> | Python | C++ |
> |---|---|
> | `class B(ABC)` + `@abstractmethod` | `struct B { virtual ~B() = default; virtual R f() = 0; };` |
> | `TypeError` on instantiation | compile error: cannot instantiate abstract class |
> | Duck typing | templates, or C++20 concepts |
> | `typing.Protocol` | a concept, or a non-owning interface reference |
> | `singledispatch` | function overloading (compile time) / visitor or `std::variant` + `std::visit` |
> | `ABC.register` | no equivalent, and you do not want one |
> | Hook method with a default | `virtual` with a body; subclass uses `override` |
>
> Things Python does not make you think about and C++ does:
>
> - **Virtual destructor.** Deleting a derived object through a `Base*` when `~Base()` is not
>   virtual is undefined behaviour. The simple rule: every polymorphic base gets
>   `virtual ~Base() = default;`.
> - **Slicing.** `Base b = derived;` copies only the base part and silently loses the
>   override. Hold polymorphic objects by reference or pointer — `std::unique_ptr<Instrument>`,
>   `const Instrument&` — never by value.
> - **`override` and `final`.** Always write `override`; it turns a typo'd signature from a
>   silent new method into a compile error. `final` on a leaf class lets the compiler
>   devirtualise.
> - **Dispatch cost.** A virtual call is an indirect jump — fine at strategy granularity,
>   measurable in a tick-handling hot loop. The usual answer there is CRTP or a
>   `std::variant` visitor, both of which resolve at compile time. Design with virtuals
>   first, measure, then flatten the paths that show up in the profile.

# Part 2 — Applied to PnL and trading logic

The design question in a PnL system: **what varies?**

- the instrument (spot, linear perp, inverse perp, dated future)
- the accounting convention (FIFO, average cost, LIFO)
- the fee schedule (flat, maker/taker, volume tiers)
- the carry (perp funding, FX rate differential)
- the mark (last, mid, weighted mid)

Each is a separate axis. That is a strong signal for **composition**: one engine that is
*given* an instrument and a fee model, not an engine subclassed per combination. Five axes
with three options each is 243 subclasses if you inherit and 15 small classes if you compose.

## 2.1 The value types

````python
from abc import ABC, abstractmethod
from collections import deque
from dataclasses import dataclass


@dataclass(frozen=True)
class Fill:
    symbol: str
    qty: float            # SIGNED: + is buy, - is sell. One convention, enforced at the edge.
    price: float
    liquidity: str = "taker"   # "maker" | "taker"
    ts: int = 0

    def __post_init__(self):
        if self.qty == 0.0:
            raise ValueError("zero-qty fill")
        if self.liquidity not in ("maker", "taker"):
            raise ValueError(f"bad liquidity flag: {self.liquidity}")
````

The signed-qty convention is worth belabouring. Half of all PnL bugs are a sign flipped at a
boundary. Pick the convention once, validate it in the constructor of the type that crosses
the boundary, and never pass an unsigned qty plus a side string any deeper than the adapter.

## 2.2 `Instrument` — where the payoff formula lives

Spot and linear perps share a formula. Inverse (coin-margined) contracts do not, and that is
exactly why the payoff belongs behind an interface rather than inline in the engine.

````python
class Instrument(ABC):
    @property
    @abstractmethod
    def pnl_ccy(self):
        # Which currency pnl() is denominated in. See the pitfall in 2.9.
        ...

    @abstractmethod
    def notional(self, qty, price):
        # Absolute exposure, used for fee calculation.
        ...

    @abstractmethod
    def pnl(self, entry, exit_price, qty):
        # PnL of a SIGNED qty opened at entry and closed at exit_price.
        ...


class SpotFX(Instrument):
    def __init__(self, pair="AUDUSD", quote_ccy="USD"):
        self.pair = pair
        self._ccy = quote_ccy

    @property
    def pnl_ccy(self):
        return self._ccy

    def notional(self, qty, price):
        return abs(qty) * price

    def pnl(self, entry, exit_price, qty):
        return qty * (exit_price - entry)


class LinearPerp(Instrument):
    def __init__(self, multiplier=1.0, quote_ccy="USDT"):
        self.multiplier = multiplier
        self._ccy = quote_ccy

    @property
    def pnl_ccy(self):
        return self._ccy

    def notional(self, qty, price):
        return abs(qty) * price * self.multiplier

    def pnl(self, entry, exit_price, qty):
        return qty * (exit_price - entry) * self.multiplier


class InversePerp(Instrument):
    # Coin-margined: 1 contract = contract_usd of notional, PnL settles in the BASE coin.
    def __init__(self, contract_usd=1.0, base_ccy="BTC"):
        self.contract_usd = contract_usd
        self._ccy = base_ccy

    @property
    def pnl_ccy(self):
        return self._ccy

    def notional(self, qty, price):
        return abs(qty) * self.contract_usd      # already USD, price-independent

    def pnl(self, entry, exit_price, qty):
        return qty * self.contract_usd * (1.0 / entry - 1.0 / exit_price)
````

````python
# Same call site, three payoff shapes. Long 100 units, 60000 -> 61000.
for inst in (SpotFX("BTCUSD"), LinearPerp(0.01), InversePerp(1.0)):
    p = inst.pnl(60_000.0, 61_000.0, 100.0)
    print(f"{type(inst).__name__:<12} pnl={p:>12,.6f} {inst.pnl_ccy}")
````

Look at the inverse number. It is small, positive, and denominated in BTC — not a bug, but an
engine that naively adds it to a USD figure produces a plausible-looking wrong answer. This is
the classic leaky abstraction: the interface unified the *shape* of the call and quietly let a
unit mismatch through. Flagged again in 2.9, and it is one of the exercises.

## 2.3 `FeeModel` — a second, independent axis

````python
class FeeModel(ABC):
    @abstractmethod
    def fee(self, notional, liquidity):
        # Positive = cost, negative = rebate.
        ...


class FlatBps(FeeModel):
    def __init__(self, bps):
        self.bps = bps

    def fee(self, notional, liquidity):
        return notional * self.bps / 1e4


class MakerTaker(FeeModel):
    def __init__(self, maker_bps, taker_bps):
        self.maker_bps = maker_bps
        self.taker_bps = taker_bps

    def fee(self, notional, liquidity):
        bps = self.maker_bps if liquidity == "maker" else self.taker_bps
        return notional * bps / 1e4


class VolumeTiered(FeeModel):
    # Stateful: an implementation is free to carry state the interface says nothing about.
    def __init__(self, tiers):
        self.tiers = sorted(tiers)          # [(volume_threshold, taker_bps), ...]
        self.volume = 0.0

    def _bps(self):
        bps = self.tiers[0][1]
        for threshold, tier_bps in self.tiers:
            if self.volume >= threshold:
                bps = tier_bps
        return bps

    def fee(self, notional, liquidity):
        bps = self._bps()
        self.volume += notional
        return notional * bps / 1e4


tiered = VolumeTiered([(0.0, 5.0), (1_000_000.0, 3.0), (10_000_000.0, 1.5)])
for n in (400_000.0, 400_000.0, 400_000.0, 9_000_000.0, 100_000.0):
    print(f"notional {n:>12,.0f}  fee {tiered.fee(n, 'taker'):>10,.2f}  "
          f"cumvol {tiered.volume:>14,.0f}")
````

`MakerTaker(maker_bps=-1.0, taker_bps=4.5)` expresses a maker rebate with no change to the
interface or to any caller. That is the test of whether an abstraction was drawn in the right
place: a new variant should not require touching anything above it.

## 2.4 `PnLEngine` — template method over the accounting convention

The sequence is invariant: charge the fee, release realized PnL against open lots, update the
position. Only the **matching rule** differs between conventions, so only that is abstract.

````python
class PnLEngine(ABC):
    def __init__(self, instrument, fee_model):
        self.instrument = instrument
        self.fee_model = fee_model
        self.position = 0.0
        self.realized = 0.0
        self.fees = 0.0
        self.funding = 0.0
        self.n_fills = 0

    # ---- concrete: identical for every convention
    def on_fill(self, fill):
        self.fees += self.fee_model.fee(
            self.instrument.notional(fill.qty, fill.price), fill.liquidity)
        self.realized += self._match(fill)
        self.position += fill.qty
        self.n_fills += 1
        return self

    def on_carry(self, cash_flow, mark, hours):
        self.funding += cash_flow.accrue(self.position, mark, hours)
        return self

    def unrealized(self, mark):
        return sum(self.instrument.pnl(entry, mark, qty) for qty, entry in self.open_lots())

    def total(self, mark):
        return self.realized + self.unrealized(mark) - self.fees + self.funding

    def summary(self, mark):
        return (f"{type(self).__name__:<20} pos {self.position:>8,.2f} | "
                f"real {self.realized:>12,.2f} | unreal {self.unrealized(mark):>12,.2f} | "
                f"fees {self.fees:>9,.2f} | fund {self.funding:>9,.2f} || "
                f"total {self.total(mark):>12,.2f}")

    # ---- abstract: the part that actually varies
    @abstractmethod
    def _match(self, fill):
        # Realized PnL released by this fill, gross of fees.
        ...

    @abstractmethod
    def open_lots(self):
        # [(signed_qty, entry_price), ...] still open.
        ...
````

````python
class FIFOEngine(PnLEngine):
    def __init__(self, instrument, fee_model):
        super().__init__(instrument, fee_model)
        self._lots = deque()

    def _match(self, fill):
        qty, realized = fill.qty, 0.0
        while qty != 0.0 and self._lots and (self._lots[0][0] > 0) != (qty > 0):
            lot_qty, lot_px = self._lots[0]
            sign = 1.0 if lot_qty > 0 else -1.0
            take = min(abs(qty), abs(lot_qty))
            closed = sign * take                       # signed qty of the lot being closed
            realized += self.instrument.pnl(lot_px, fill.price, closed)
            if abs(abs(lot_qty) - take) < 1e-12:
                self._lots.popleft()
            else:
                self._lots[0] = (lot_qty - closed, lot_px)
            qty += closed                              # incoming shrinks toward zero
        if abs(qty) > 1e-12:
            self._lots.append((qty, fill.price))       # remainder opens a new lot
        return realized

    def open_lots(self):
        return list(self._lots)


class AverageCostEngine(PnLEngine):
    def __init__(self, instrument, fee_model):
        super().__init__(instrument, fee_model)
        self._qty = 0.0
        self._avg = 0.0

    def _match(self, fill):
        qty, realized = fill.qty, 0.0
        if self._qty != 0.0 and (self._qty > 0) != (qty > 0):
            take = min(abs(qty), abs(self._qty))
            closed = (1.0 if self._qty > 0 else -1.0) * take
            realized = self.instrument.pnl(self._avg, fill.price, closed)
            self._qty -= closed
            qty += closed
        if abs(qty) > 1e-12:
            if abs(self._qty) < 1e-12:
                self._avg = fill.price
            else:
                self._avg = (self._avg * self._qty + fill.price * qty) / (self._qty + qty)
            self._qty += qty
        return realized

    def open_lots(self):
        return [(self._qty, self._avg)] if abs(self._qty) > 1e-12 else []
````

Both engines are ~20 lines because everything they share — fees, funding, marking, totals — is
in the base. Adding LIFO or highest-cost-first means writing `_match` and `open_lots` and
nothing else.

## 2.5 Same fills, two conventions

````python
FILLS = [
    Fill("BTC", +2.0, 60_000.0, "maker"),
    Fill("BTC", +3.0, 61_000.0, "taker"),
    Fill("BTC", -4.0, 62_000.0, "taker"),   # crosses both lots under FIFO
    Fill("BTC", +1.0, 61_500.0, "maker"),
]
MARK = 61_800.0


def run(engine_cls, fills=FILLS, mark=MARK):
    eng = engine_cls(LinearPerp(1.0), MakerTaker(maker_bps=-1.0, taker_bps=4.5))
    for f in fills:
        eng.on_fill(f)
    print(eng.summary(mark))
    return eng


fifo = run(FIFOEngine)
avg = run(AverageCostEngine)
print()
print(f"realized differs by {fifo.realized - avg.realized:,.2f}")
print(f"total differs by    {fifo.total(MARK) - avg.total(MARK):,.2f}")
print("open lots FIFO:", [(round(q, 4), round(p, 2)) for q, p in fifo.open_lots()])
print("open lots AVG :", [(round(q, 4), round(p, 2)) for q, p in avg.open_lots()])
````

Realized and unrealized differ; **total is identical**. That is the invariant worth
remembering: accounting convention moves PnL between buckets, it does not create or destroy
it. If a convention change moves your total while the position is unchanged, you have a bug,
not a convention.

Flatten the book completely and even the split converges:

````python
FLAT = FILLS + [Fill("BTC", -2.0, 61_800.0, "taker")]   # ends at zero position
for cls in (FIFOEngine, AverageCostEngine):
    e = cls(LinearPerp(1.0), MakerTaker(-1.0, 4.5))
    for f in FLAT:
        e.on_fill(f)
    print(f"{cls.__name__:<20} pos={e.position:.2f} realized={e.realized:>10,.2f} "
          f"unreal={e.unrealized(MARK):>6,.2f} fees={e.fees:>8,.2f} "
          f"total={e.total(MARK):>10,.2f}")
````

### The invariant has a hidden assumption

"Total is convention-independent" held above because `LinearPerp` pays off linearly in price.
Run the same fills through `InversePerp` (contracts instead of BTC, so scale the quantities up):

````python
INV_FILLS = [Fill(f.symbol, f.qty * 10_000, f.price, f.liquidity) for f in FILLS]

for cls in (FIFOEngine, AverageCostEngine):
    e = cls(InversePerp(1.0), FlatBps(0.0))
    for f in INV_FILLS:
        e.on_fill(f)
    print(f"{cls.__name__:<20} realized={e.realized:.8f}  unreal={e.unrealized(MARK):.8f}  "
          f"total={e.total(MARK):.8f} BTC")
````

Now the totals disagree. Neither `FIFOEngine` nor `InversePerp` is wrong on its own.
`AverageCostEngine` averages entry **prices** arithmetically, which silently assumes a payoff
linear in price, and inverse contracts pay off linearly in `1/price`. The assumption lives in
the engine, the fact that breaks it lives in the instrument, and the interface between them
carries neither. No ABC or type check catches that; only a test that runs every engine against
every instrument does. Deciding where the averaging rule should live is exercise 8.

## 2.6 Carry — perp funding and FX rate differential behind one interface

Different markets, same shape: a cash flow proportional to position, notional and elapsed
time. Worth noting that this is the same abstraction whether the desk is quoting perps or
holding an FX carry position.

````python
class CashFlow(ABC):
    @abstractmethod
    def accrue(self, position, mark, hours):
        # Cash flow over `hours`. Positive = received.
        ...


class PerpFunding(CashFlow):
    # Longs pay shorts when the rate is positive.
    def __init__(self, rate_per_8h):
        self.rate_per_8h = rate_per_8h

    def accrue(self, position, mark, hours):
        return -position * mark * self.rate_per_8h * (hours / 8.0)


class FXCarry(CashFlow):
    # Annualised rate differential, accrued act/365.
    def __init__(self, rate_diff_annual):
        self.rate_diff_annual = rate_diff_annual

    def accrue(self, position, mark, hours):
        return position * mark * self.rate_diff_annual * (hours / (24.0 * 365.0))


e = FIFOEngine(LinearPerp(1.0), MakerTaker(-1.0, 4.5))
for f in FILLS:
    e.on_fill(f)
e.on_carry(PerpFunding(0.0001), MARK, hours=8)    # long 2 -> pays
e.on_carry(PerpFunding(-0.0002), MARK, hours=8)   # negative rate -> receives
print(e.summary(MARK))

fx = FIFOEngine(SpotFX("AUDUSD"), FlatBps(0.2))
fx.on_fill(Fill("AUDUSD", +1_000_000.0, 0.6600))
fx.on_carry(FXCarry(0.0125), 0.6600, hours=24 * 30)
print(fx.summary(0.6650))
````

## 2.7 The contract test — one test, every implementation

This is the real payoff of the base class, and the answer to the Liskov problem in 1.12. The
properties are written **once against the interface** and run against every subclass, present
and future.

````python
def check_engine_contract(engine_cls, verbose=True):
    inst, fees = LinearPerp(1.0), FlatBps(0.0)      # zero fees isolates the accounting
    mark = 105.0

    # P1: a single open fill realizes nothing
    e = engine_cls(inst, fees)
    e.on_fill(Fill("X", +1.0, 100.0))
    assert abs(e.realized) < 1e-9, f"{engine_cls.__name__}: realized on an opening fill"

    # P2: round trip realizes exactly the spread and leaves no open lots
    e.on_fill(Fill("X", -1.0, 110.0))
    assert abs(e.realized - 10.0) < 1e-9, f"{engine_cls.__name__}: realized={e.realized}"
    assert abs(e.position) < 1e-9
    assert e.open_lots() == [] or all(abs(q) < 1e-12 for q, _ in e.open_lots())

    # P3: open lots must reconcile to the reported position
    e = engine_cls(inst, fees)
    for f in FILLS:
        e.on_fill(f)
    lot_sum = sum(q for q, _ in e.open_lots())
    assert abs(lot_sum - e.position) < 1e-9, f"{engine_cls.__name__}: lots {lot_sum} != pos {e.position}"

    # P4: a flat book has nothing left unrealized
    e = engine_cls(inst, fees)
    for f in FLAT:
        e.on_fill(f)
    assert abs(e.total(mark) - e.realized) < 1e-9, "flat book: total must equal realized"

    # P5: shorts work symmetrically
    e = engine_cls(inst, fees)
    e.on_fill(Fill("X", -2.0, 100.0))
    e.on_fill(Fill("X", +2.0, 90.0))
    assert abs(e.realized - 20.0) < 1e-9, f"{engine_cls.__name__}: short round trip {e.realized}"

    # P6: total must equal the cash-flow identity, whatever the convention.
    #     For a linear instrument: sum over fills of qty * (mark - fill price).
    #     P1-P5 check internal consistency; this is the one that checks the number is RIGHT.
    for fills in (FILLS, FLAT):
        e = engine_cls(inst, fees)
        for f in fills:
            e.on_fill(f)
        truth = sum(f.qty * (MARK - f.price) for f in fills)
        assert abs(e.total(MARK) - truth) < 1e-6, \
            f"{engine_cls.__name__}: total {e.total(MARK):,.2f} != cash identity {truth:,.2f}"

    if verbose:
        print(f"{engine_cls.__name__:<20} PASS")


for cls in (FIFOEngine, AverageCostEngine):
    check_engine_contract(cls)
````

````python
# A convention that violates the contract is caught immediately, not in production.
class NaiveEngine(PnLEngine):
    # Marks everything against the LAST fill price — a common shortcut, and wrong.
    def __init__(self, instrument, fee_model):
        super().__init__(instrument, fee_model)
        self._last = 0.0
        self._qty = 0.0

    def _match(self, fill):
        realized = 0.0
        if self._qty != 0.0 and (self._qty > 0) != (fill.qty > 0):
            realized = self.instrument.pnl(self._last, fill.price,
                                           min(abs(fill.qty), abs(self._qty)) *
                                           (1.0 if self._qty > 0 else -1.0))
        self._qty += fill.qty
        self._last = fill.price
        return realized

    def open_lots(self):
        return [(self._qty, self._last)] if abs(self._qty) > 1e-12 else []


try:
    check_engine_contract(NaiveEngine)
except AssertionError as e:
    print("AssertionError:", e)
````

`NaiveEngine` sails through P1–P5: it realizes nothing on open, gets a single round trip
right, keeps lots reconciled to position and handles shorts. It fails only P6, once
there are several fills at different prices. That is typical. Checks of internal consistency
are easy to pass while the number is wrong, so every contract test needs at least one property
tied to an independent calculation of the truth.

Write this once per interface and it keeps paying. It is also the cheapest way to review
someone else's new implementation: if it passes the contract test, the conversation is about
performance and style rather than correctness.

## 2.8 Strategy as a template method: quoter / hedger

The risk pipeline is fixed. What varies is how a quote is priced. Two implementations below:
a symmetric quoter, and one that skews on the basis between two venues.

````python
@dataclass
class Book:
    bid: float
    ask: float

    @property
    def mid(self):
        return (self.bid + self.ask) / 2.0


class Quoter(ABC):
    def __init__(self, engine, max_position, size):
        self.engine = engine
        self.max_position = max_position
        self.size = size

    # ---- concrete: every quoter goes through the same gates, in the same order
    def on_book(self, book, **ctx):
        quotes = self.price(book, **ctx)
        quotes = self.skew_for_inventory(quotes, book)
        return self.apply_position_limit(quotes)

    def skew_for_inventory(self, quotes, book):
        # Default hook: lean the quote away from existing inventory.
        lean = (self.engine.position / self.max_position) * (book.ask - book.bid) * 0.5
        return {"bid": quotes["bid"] - lean, "ask": quotes["ask"] - lean}

    def apply_position_limit(self, quotes):
        pos = self.engine.position
        out = dict(quotes)
        if pos >= self.max_position:
            out["bid"] = None                 # no more buying
        if pos <= -self.max_position:
            out["ask"] = None
        return out

    # ---- abstract
    @abstractmethod
    def price(self, book, **ctx):
        ...


class SymmetricQuoter(Quoter):
    def __init__(self, engine, max_position, size, half_spread):
        super().__init__(engine, max_position, size)
        self.half_spread = half_spread

    def price(self, book, **ctx):
        return {"bid": book.mid - self.half_spread, "ask": book.mid + self.half_spread}


class BasisQuoter(Quoter):
    # Quotes venue A, skewed by the basis to venue B, where the hedge would be done.
    def __init__(self, engine, max_position, size, half_spread, skew_frac=0.5):
        super().__init__(engine, max_position, size)
        self.half_spread = half_spread
        self.skew_frac = skew_frac

    def price(self, book, **ctx):
        hedge = ctx["hedge_book"]
        basis = hedge.mid - book.mid
        centre = book.mid + self.skew_frac * basis
        return {"bid": centre - self.half_spread, "ask": centre + self.half_spread}


def show(tag, q):
    fmt = lambda x: "  --  " if x is None else f"{x:,.2f}"
    print(f"{tag:<28} bid {fmt(q['bid'])}   ask {fmt(q['ask'])}")


eng = FIFOEngine(LinearPerp(1.0), MakerTaker(-1.0, 4.5))
venue_a = Book(59_995.0, 60_005.0)
venue_b = Book(60_020.0, 60_030.0)

sym = SymmetricQuoter(eng, max_position=5.0, size=1.0, half_spread=8.0)
bas = BasisQuoter(eng, max_position=5.0, size=1.0, half_spread=8.0)

show("flat, symmetric", sym.on_book(venue_a))
show("flat, basis-skewed", bas.on_book(venue_a, hedge_book=venue_b))

eng.on_fill(Fill("BTC", +4.0, 60_000.0, "maker"))
show("long 4, symmetric", sym.on_book(venue_a))
show("long 4, basis-skewed", bas.on_book(venue_a, hedge_book=venue_b))

eng.on_fill(Fill("BTC", +1.0, 60_000.0, "maker"))
show("long 5 (at limit)", sym.on_book(venue_a))
````

The position limit is unskippable because no subclass writes `on_book`. A new quoter is one
method. That is the same argument as 1.6, but it is the version that matters: the invariant
being protected is a risk control.

Note the `**ctx` on `price`. It is a small concession — `BasisQuoter` needs an input
`SymmetricQuoter` does not — and it is a warning sign worth watching. If half the subclasses
need context the base knows nothing about, the interface was drawn at the wrong level and a
richer market-state object is the better shape.

## 2.9 Pitfalls specific to PnL and trading code

**Floats for money.** Every number above is a float. It is fine for a research notebook and
wrong for anything that reconciles against a venue. Use integer minor units (satoshis, cents)
or `decimal.Decimal` at the ledger boundary, and keep floats for research.

**Mixed currencies behind one interface.** `InversePerp.pnl` returns BTC; the others return
quote currency. `total()` adds them without complaint. Either the interface carries units
(`pnl_ccy` is a start, but nothing enforces it) or a conversion is mandatory at the boundary.
Making this impossible-by-construction is an exercise below.

**Abstracting before the second implementation exists.** An ABC with one subclass is a class
with extra steps. Wait for the second real case; the right seams are obvious then and guessed
badly before.

**Too many abstract methods.** If a new implementation has to write eight methods, nobody
writes new implementations. Push everything you can into concrete base methods and hooks with
defaults, leave the smallest possible abstract surface.

**Inheritance where composition belongs.** `FIFOPerpMakerTakerEngine` is the disease.
Instrument, fee model and carry are constructor arguments precisely so they multiply instead
of combining into a class per permutation.

**The base that assumes a market structure.** A `PnLEngine` that assumes continuous two-sided
quoting breaks on an auction, and one that assumes a single settlement currency breaks on the
first coin-margined contract. Abstract bases encode assumptions; the cost of a wrong one
scales with the number of subclasses.

**ABCs do not check semantics.** Repeating 1.12 because it is the one that actually loses
money: the contract test is the enforcement mechanism, not the base class.

> ### C++ mapping — Part 2
>
> ```cpp
> class Instrument {
> public:
>     virtual ~Instrument() = default;
>     virtual double notional(double qty, double price) const = 0;
>     virtual double pnl(double entry, double exit, double qty) const = 0;
> };
>
> class LinearPerp final : public Instrument {
>     double mult_;
> public:
>     explicit LinearPerp(double mult) : mult_(mult) {}
>     double notional(double qty, double price) const override {
>         return std::abs(qty) * price * mult_;
>     }
>     double pnl(double entry, double exit, double qty) const override {
>         return qty * (exit - entry) * mult_;
>     }
> };
>
> class FeeModel {
> public:
>     virtual ~FeeModel() = default;
>     virtual double fee(double notional, Liquidity liq) = 0;   // NOT const: tiers mutate
> };
>
> class PnLEngine {
> public:
>     PnLEngine(const Instrument& inst, FeeModel& fees)   // non-owning refs
>         : inst_(inst), fees_(fees) {}
>     virtual ~PnLEngine() = default;
>
>     void on_fill(const Fill& f) {          // template method: non-virtual on purpose
>         fees_paid_ += fees_.fee(inst_.notional(f.qty, f.price), f.liquidity);
>         realized_  += match(f);
>         position_  += f.qty;
>     }
> protected:
>     virtual double match(const Fill& f) = 0;
>     const Instrument& inst_;
>     FeeModel&         fees_;
>     double position_{}, realized_{}, fees_paid_{};
> };
> ```
>
> Points that have no Python analogue:
>
> - **Ownership.** `const Instrument&` is non-owning, so the instrument must outlive the
>   engine. If the engine should own it, `std::unique_ptr<Instrument>`. Decide deliberately —
>   in Python the GC hides this question entirely.
> - **`const` correctness.** `pnl` and `notional` are `const`; `fee` is not, because the
>   volume-tiered model mutates. So the engine holds `FeeModel&`, not `const FeeModel&`, since
>   a non-const method cannot be called through a const reference. One stateful
>   implementation forced that decision onto the whole interface. Python never raises the
>   question; in C++ it is part of the contract.
> - **Reference members** make the engine non-assignable (`engine_a = engine_b` will not
>   compile). Usually what you want for an engine; worth knowing before you put one in a
>   `std::vector`.
> - **Non-virtual template method.** `on_fill` is deliberately *not* virtual — the sequence is
>   the guarantee. The virtual hole is `match`, and it is `protected`, not `public`. This is
>   the non-virtual interface idiom, and it states the intent more strongly than Python can.
> - **`final` on leaf classes** lets the compiler devirtualise and is free.
> - **Hot path.** If `on_fill` runs per tick, benchmark the virtual `match` call. CRTP
>   (`template <class Derived> class PnLEngine`) or `std::variant<FIFO, AverageCost>` plus
>   `std::visit` both remove the indirection at the cost of losing runtime pluggability.
>   Do this after measuring, not before.
> - **Contract tests port directly.** A `TEST_P` parameterised over engine types in GoogleTest
>   is the same idea as `check_engine_contract`, and it is the artefact most likely to get your
>   code accepted by whoever maintains the strategy codebase.
